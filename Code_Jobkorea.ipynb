```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
import pandas as pd
import time

# 크롬 옵션 설정
options = Options()
options.add_argument("--headless")
options.add_argument("--disable-gpu")
options.add_argument("--window-size=1920x1080")

# 드라이버 실행
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# 검색 결과 페이지 접속
search_url = "https://www.jobkorea.co.kr/Search/?stext=데이터분석"
driver.get(search_url)
time.sleep(3)

# 결과 저장용 리스트
job_list = []


for i in range(2, 20):  
    try:
        xpath = f'//*[@id="jk-«R1h7alqdb»-content-recruit"]/div/div/div[{i}]/div'
        title_element = driver.find_element(By.XPATH, xpath)
        lines = title_element.text.split('\n')

        # 링크 추출
        link_xpath = f'//*[@id="jk-«R1h7alqdb»-content-recruit"]/div/div/div[{i}]/div//a'
        link_element = driver.find_element(By.XPATH, link_xpath)
        link_url = link_element.get_attribute("href")


        if len(lines) >= 4:
            company_name = lines[2]
            job_title = lines[3]
            details = ', '.join(lines[4:-1])  # 줄바꿈 대신 쉼표로 연결

            job_list.append({
                "Site": "Job_Korea",
                "Col_Company": company_name,
                "Col_Recruit": job_title,
                "Col_detail": details,
                "Col_url": link_url
            })
        else:
            print(f"[{i}] 정보 없음")
    except Exception as e:
        print(f"[{i}] 실패")

# 드라이버 종료
driver.quit()

# DataFrame 생성
jobkorea_df = pd.DataFrame(job_list)


#display(jobkorea_df)
```

```python
jobkorea_df.to_csv("data_tmp/data_Jobkorea.csv", index=False, encoding="utf-8-sig")
```
